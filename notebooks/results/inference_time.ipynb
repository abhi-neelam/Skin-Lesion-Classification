```python
import pandas as pd
import seaborn as sns
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt
import os
```

```python
sns.set_theme(rc={"figure.dpi": 100}) # Set DPI to 300 for high resolution
```

```python
sns.set_theme(style="whitegrid")
sns.color_palette("bright")
```

```python
d = {'Classifier': ['XGBoost', 'LightGBM', 'SVM', 'LR', 'XGBoost', 'LightGBM', 'SVM', 'LR'], 'Inference Time (s)': [0.38, 0.05, 30.05, 0.11, 0.39, 0.05, 37.5, 0.13], 'Dataset': ['ISIC 2019', 'ISIC 2019', 'ISIC 2019', 'ISIC 2019', 'ISIC-Derm-8', 'ISIC-Derm-8', 'ISIC-Derm-8', 'ISIC-Derm-8']}
df = pd.DataFrame(data=d)
fig, axes = plt.subplots(1,1,figsize=(7,6))
ax = sns.barplot(df, x="Classifier", y="Inference Time (s)", hue='Dataset', palette='bright')
for container in ax.containers:
    ax.bar_label(container)
```
