```python
import pandas as pd
import seaborn as sns
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt
import os
```

```python
sns.set_theme(rc={"figure.dpi": 100})
```

```python
sns.set_theme(style="whitegrid")
sns.color_palette("husl")
```

```python
d = {'Architecture': ['MobileNet-v3 Large', 'FasterNet-T1', 'shViT-s4', 'MobileNet-v3 Large', 'FasterNet-T1', 'shViT-s4'], 'F1-Score': [86.47, 85.01, 85.53, 88.29, 88.97, 87.58], 'Augmentation Procedure': ['Standard', 'Standard', 'Standard', 'On-The-Fly', 'On-The-Fly', 'On-The-Fly']}
df = pd.DataFrame(data=d)
fig, axes = plt.subplots(1,1,figsize=(7,6))
ax = sns.barplot(df, x="Architecture", y="F1-Score", hue='Augmentation Procedure', palette='husl')
for container in ax.containers:
    ax.bar_label(container)
```
